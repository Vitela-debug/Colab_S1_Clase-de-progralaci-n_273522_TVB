# Actividad — Comparar el clima entre varias ciudades con Open-Meteo (Sesión 8)

En el Bloque 2 consultaste el pronóstico por hora de Ciudad Juárez con Open-Meteo. En esta actividad usas
la misma API, pero de otra forma: en vez de un pronóstico por hora de una sola ciudad, se consulta el
**clima actual** de **varias ciudades distintas** y se construye una tabla comparativa.

**Endpoint:** `https://api.open-meteo.com/v1/forecast`
**Parámetro clave:** `current_weather=true` — regresa un solo valor actual (no una lista por hora) dentro
de `datos["current_weather"]`, con `temperature`, `windspeed`, `weathercode` y `time`.

**Instrucciones:**

1. Define una lista `ciudades` con al menos tres ciudades (nombre, latitud, longitud), distintas a
   Ciudad Juárez.
2. Para cada ciudad, consulta el endpoint con `current_weather=true`, usando `timeout` y manejo de
   errores como en el Bloque 2.
3. De cada respuesta, extrae la temperatura y la velocidad del viento actuales.
4. Estructura la información como una lista de diccionarios llamada `info_clima`.
5. Convierte `info_clima` en un DataFrame llamado `df_clima_ciudades`.
6. Verifica el resultado.

**Paso 1 — Define las ciudades a consultar.**

```python
ciudades = [
    {"nombre": "Chihuahua", "latitud": 28.63, "longitud": -106.09}, #Pre cargada en el archivo
    {"nombre": "Monterrey", "latitud": 25.69, "longitud": -100.32}, #Pre cargada en el archivo
    {"nombre": "Guadalajara", "latitud": 20.66, "longitud": -103.35}, #Pre cargada en el archivo
    {"nombre": "Tijuana", "latitud": 32.51, "longitud": -117.03}, #Agregada para el ejercio
    {"nombre": "Mérida", "latitud": 20.96, "longitud": -89.59}, #Agregada para el ejercio
    {"nombre": "Puebla", "latitud": 19.04, "longitud": -86.85}, #Agregada para el ejercio
]  # puedes cambiar o agregar ciudades

ciudades
```

---
---
* Para el paso 1 el código define una lista de diccionarios, donde cada diccionario almacena el nombre, la latitud y la longitud de una ciudad, en este caso se opto por dejar las 3 ciudades ya cargadas y agregar otras 3 para que sea interesante trabajar con los datos.
---
---

**Pasos 2 a 4 — Consultar cada ciudad y estructurar la información.**

```python
import requests

info_clima = [] #Lista vacia

for ciudad in ciudades:
    endpoint = "https://api.open-meteo.com/v1/forecast" #direccion de la API
    parametros = {
        "latitude": ciudad["latitud"], #parametros que se van a enviar
        "longitude": ciudad["longitud"], #parametros que se van a enviar
        "current_weather": True, #current_weather: True le dice a la API que quieres obtener el clima actual.
    }

    # Completa: envía la solicitud con timeout, revisa status_code con raise_for_status()
    try:
        respuesta = requests.get(endpoint, params=parametros, timeout=30) #Paso 2 Consulta el endpoint para cada ciudad
        respuesta.raise_for_status()  # revisa si hubo un error HTTP
        datos = respuesta.json()
        clima_actual = datos["current_weather"] #Para acceder a los datos del clima actual

        # Completa: extrae temperature y windspeed de clima_actual
        info_clima.append({ #Paso 4 Crea la lista de diccionarios info_clima y se agrega con .append
            "ciudad": ciudad["nombre"],
            "temperatura": clima_actual["temperature"], #Para obtener la temperatura (Paso 3)
            "viento": clima_actual["windspeed"], #Para obtener la velocidad del viento (Paso 3)
        })

    except requests.RequestException as error:
        print(f"No fue posible consultar {ciudad['nombre']}:", error) #Manejo de errores

info_clima #Para mostrar la lista info clima
```

---
---
* En el paso 2 el código consulta una API utilizando la librería requests y el método requests.get().
  * Se utiliza un ciclo for para consultar cada ciudad de la lista y se envían como parámetros su latitud y longitud para solicitar el clima actual.
  * Timeout=30 se usa para establecer un tiempo máximo de espera de 30 segundos
  * raise_for_status() se utiliza para verificar si la respuesta de la API contiene algún error.
* El código del paso 3 por su parte utiliza la respuesta de la API y accede a los datos mediante un diccionario.
  *  respuesta.json() para convertir la respuesta a un formato que Python puede manejar
  * datos["current_weather"] para acceder al clima actual y mediante las claves "temperature" y "windspeed", se extraen la temperatura actual y la velocidad del viento de cada ciudad.
* El código del paso 4 utiliza la lista info_clima y el método .append() para guardar los datos obtenidos. Cada ciudad se almacena como un diccionario que contiene su nombre, temperatura y velocidad del viento. De esta manera, se va formando una lista de diccionarios con la información del clima de todas las ciudades.

**Paso 5 — Construir el DataFrame.**

```python
# Completa: convierte info_clima en un DataFrame llamado df_clima_ciudades
import pandas as pd
df_clima_ciudades = pd.DataFrame(info_clima) #Paso 5 Crea un DataFrame a partir de la lista info_clima
df_clima_ciudades
```
