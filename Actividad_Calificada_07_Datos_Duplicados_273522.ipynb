### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Escribe tu código aquí
hoteles_duplicados = hoteles.copy()
hoteles_duplicados['es_duplicado'] = hoteles_duplicados.duplicated(keep='first') #para identificar duplicados

filas_duplicadas = hoteles_duplicados.groupby('hotel')['es_duplicado'].agg('sum') #se agrupan por hotel y se suman las true de si es duplicado
filas_hoteles = hoteles_duplicados.groupby('hotel')['es_duplicado'].agg('count') #se agrupan por hotel y se cuentan las filas
Porcentajes = (filas_duplicadas/ filas_hoteles) * 100 #se calcula el porcejtaje de cada grupo

print('Filas duplicadas por hotel')
print(filas_duplicadas)
print('Porcentajes')
print(Porcentajes)
```

---
---
* El código del ejercicio 1.3 sirve para encontrar las filas duplicadas de cada hotel y saber qué porcentaje representan del total de reservas. Primero hace una copia de los datos y marca cuáles filas están repetidas con `hoteles_duplicados.duplicated(keep='first')`, después agrupa la información por hotel y cuenta los duplicados con `hoteles_duplicados.groupby('hotel')['es_duplicado'].agg('sum')`. Luego cuenta todas las reservas de cada hotel con hoteles_duplicados.groupby('hotel')['es_duplicado'].agg('count') y calcula el porcentaje de duplicados por hotel.

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Escribe tu código aquí
tamaño_original = hoteles.groupby('hotel').size() #Agrupamiento por hoteles y ver el tamaño
medias_original = hoteles.groupby('hotel')[['is_canceled', 'adr']].mean() #sacar las medias de las dos columnas por grupos
resumen_original = pd.concat([tamaño_original.rename('Tamaño'),medias_original], axis=1).reset_index() #juntar la info en un solo Dataframe y inciar de nuevo a numerar

#Mismo proceso pero para hoteles_limpio
tamaño_limpio = hoteles_limpio.groupby('hotel').size()
medias_limpio = hoteles_limpio.groupby('hotel')[['is_canceled', 'adr']].mean()
resumen_limpio = pd.concat([tamaño_limpio.rename('Tamaño'),medias_limpio], axis=1).reset_index()

print('resumen_original')
print(resumen_original)
print('resumen_limpio')
print(resumen_limpio)
```

---
---
* El código sirve para resumir y comparar la información de las reservas por tipo de hotel. Primero, `groupby('hotel')` agrupa los datos en City Hotel y Resort Hotel,  `.mean()` calcula el promedio de is_canceled y adr y mientras que `.size()` cuenta cuántas reservas hay en cada hotel. `rename()` cambia el nombre de la cantidad de reservas a Tamaño,` pd.concat(..., axis=1) `une los resultados en columnasen un solo DataFrame y `reset_index() ` convierte hotel nuevamente en una columna. En conjunto, obtenemos un resumen con el número de reservas, la tasa de cancelación y la tarifa diaria promedio de cada hotel para el caso de datos originales y con la limpieza.

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Escribe tu código aquí
resumen_original['version'] = 'original' #Se agrega la columna de versión original
resumen_limpio['version'] = 'sin duplicados' #e agrega la columna de versión sin duplicados
comparacion = pd.concat([resumen_original, resumen_limpio], ignore_index=True) #Se unen los df con pd.condat y  ignorar y crear nuevos indices
comparacion = comparacion.sort_values(by='hotel') #Se agrupan por hotel de nuevo para que queden juntos
comparacion.reset_index(drop=True, inplace=True) #Se vuelve a acomodar el indice
comparacion
```

En general permite ver cómo cambian las reservas, cancelaciones y tarifas antes y después de eliminar los duplicados.
