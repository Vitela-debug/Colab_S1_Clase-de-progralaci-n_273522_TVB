## Ejercicio 8 — Síntesis: resumen por estación (15 pts)

Usando `map()` y/o comprensión de listas, genera una lista de cadenas de resumen, una por estación, que combine:

- El identificador de estación extraído en el Ejercicio 1.
- El promedio de temperatura corregida de esa estación (Ejercicio 7), formateado a un decimal.
- Su clasificación predominante (`'alta'` o `'normal'`) — usa `np.unique()` sobre las etiquetas de esa estación (del Ejercicio 6) para encontrar cuál aparece con mayor frecuencia entre las dos.

Formato esperado por cadena: `"EST-01: 23.4°C promedio — clasificación predominante: normal"`

Finalmente, ordena la lista de resúmenes de mayor a menor promedio de temperatura usando `sorted()` con `key` y `lambda` (como en la Sesión 4/5), e imprime el resultado ordenado.

````python
# Tu código aquí


# Extraer estación y temperatura (Ejercicio 1)
extraidos = [(re.search(r'EST-\d+', lectura).group(), re.search(r'Temp:\s*([\d.]+)C', lectura).group(1)) #\d+ Busca uno o mas digitos, \s* para cero o más espacios, .]+) punto decimal y mas caracteres
    for lectura in lecturas_crudas] #para la lista de compresion

# Matriz corregida (Ejercicio 6)
matriz_corregida = matriz_temp + descalibre.reshape(4, 1)

# Clasificación (Ejercicioo 6)
etiquetas = np.where(matriz_corregida > 25, 'alta', 'normal')

# Promedio por estación (Ejercicio 7)
promedio_estacion = np.mean(matriz_corregida, axis=1)

# Obtener los identificadores de estación sin repetir
estaciones = list(dict.fromkeys([dato[0] for dato in extraidos])) #Se usa la comprension de lista para solo tomar el primer valor de la tupla ('Est-01,'23') de extraidos

# Crear resumen que luego se ordenara
resumenes = []

for i, estacion in enumerate(estaciones):
    etiquetas_estacion = etiquetas[i] # Etiquetas de la estación
    valores, frecuencias = np.unique(etiquetas_estacion,return_counts=True) # Verifica cuántas veces aparece 'alta' y 'normal' en la estación
    clasificacion_predominante = valores[frecuencias == frecuencias.max()][0] # Busca la clasificación que tiene la frecuencia más alta y guarda como la clasificación predominante
    resumen = (f"{estacion}: {promedio_estacion[i]:.1f}°C promedio — "f"clasificación predominante: {clasificacion_predominante}") #Genera el resumen de lo que se obtuvo anteriormente (:.1f formato de número decimal (float) una cifra despues del punto)
    resumenes.append(resumen)

# Ordenar de mayor a menor promedio
resumenes_ordenados = sorted(resumenes, key=lambda x: float(x.split(": ")[1].split("°C")[0]),reverse=True)

print(resumenes_ordenados)
````
