## Ejercicio 1 — `enumerate()` (10 puntos)

Construye la lista `reporte_ciudades`, donde cada elemento sea una cadena con el formato `"Ciudad N: nombre"`, numerando **a partir de 1** (usa el parámetro `start` de `enumerate`). Recorre `ciudades`.

EL código crea una lista llamada reporte_ciudades donde se muestra el número de cada estación {i} y el nombre de la ciudad, enumerate sirve para poner un índice que empieza desde 1 (se utliza el ciclo for para recorrer los datos de estaciones).

## Ejercicio 2 — `zip()` + `dict()` (10 puntos)

Construye `temperatura_por_ciudad`, un diccionario que empareje cada ciudad con su temperatura, usando `dict(zip(...))` sobre `ciudades` y `temperaturas`.


---
---
EL código del ejercicio 2 toma de la lista estaciones los nombres de las ciudades y las temperaturas, y los guarda por separado en dos listas. Después, con zip, junta cada ciudad con su temperatura y con dict crea un diccionario.
---
---

````python
# ESCRIBE TU CÓDIGO AQUÍ
ciudades = [e["nombre"] for e in estaciones] #Se extrae la informacion de la lista para ciudades y se guarda
temperaturas = [e["temperatura"] for e in estaciones] #Se extrae la informacion de la lista para temperatura y se guarda

temperatura_por_ciudad = dict(zip(ciudades, temperaturas)) #Emparejamos cada ciudad con su temperatura (zip) y creamos un diccionario dict


print(temperatura_por_ciudad)
````

## Ejercicio 3 — `zip()` + `enumerate()` (10 puntos)

Usando el patrón `enumerate(zip(...))`, imprime una línea por ciudad con el formato exacto:

```
1. Juárez: 31.2°C
2. Chihuahua: 28.5°C
...
```

(numeración desde 1). No necesitas guardar el resultado en una variable — solo imprime.


---
---
En el ejercio 3 el código recorre las listas de ciudades y temperaturas al mismo tiempo. zip sirve para juntar cada ciudad con su temperatura y enumerate agrega un número que empieza desde 1, y print muestra cada ciudad con su temperatura en una línea diferente
---
---


````python
# ESCRIBE TU CÓDIGO AQUÍ
for i, (ciudad, temperatura) in enumerate(zip(ciudades, temperaturas), start=1): #enumerate para recorrer la lista y zip para hacer el emparejamiento de las listas previamiente generadas
    print(f"{i}. {ciudad}: {temperatura}°C") #Print dentro del for para imprimir sin variable
````

## Ejercicio 4 — `sorted()` (10 puntos)

Construye `orden_por_temperatura`: una lista de tuplas `(temperatura, ciudad)`, ordenada de **menor a mayor temperatura**.

Pista: no necesitas `key` para este ejercicio — piensa en qué compara Python primero cuando ordena una lista de tuplas, y en qué fue lo primero que le diste a `zip`.


---
---
Para el ejercicio 4 el código toma de la lista estaciones la temperatura y el nombre de cada ciudad y los guarda juntos en forma de tuplas. Después, sorted ordena esas tuplas de menor a mayor tomando primero como referencia la temperatura, que es el primer dato asignado
---
---

````python
# ESCRIBE TU CÓDIGO AQUÍ
orden_por_temperatura = sorted([(e["temperatura"], e["nombre"]) for e in estaciones]) #primero se generan las tuplas y luego se ordenan con el primer elemento que es temperatura


print(orden_por_temperatura)
````

## Ejercicio 5 — `reversed()` (10 puntos)

Usando `orden_por_temperatura` del ejercicio anterior, construye `orden_descendente`: la misma lista de tuplas, pero de mayor a menor temperatura — **sin volver a ordenar**, solo invirtiendo el resultado que ya tienes.


---
---
En el ejercicio 5 el código toma la lista que ya estaba ordenada por temperatura usa reversed para inviertir su orden (guarda el resultado en una nueva lista llamada orden_descendente)
---
---


````python
# ESCRIBE TU CÓDIGO AQUÍ
orden_descendente = [e for e in reversed(orden_por_temperatura)] #reversed para invertir el orden de la lista que se esta generando


print(orden_descendente)
````

## Ejercicio 6 — Comprensión de listas (10 puntos)

Construye `ciudades_calidas`: una lista con el `nombre` de las estaciones cuya `temperatura` sea mayor a 27, usando una comprensión de listas sobre `estaciones`.


---
---
En el ejercicio 6 el código crea una nueva lista con los nombres de las ciudades que tienen una temperatura mayor a 27 grados (utilizando un if).  Recorriendo todas las estaciones mediante una compresion de listas.
---
---


````python
# ESCRIBE TU CÓDIGO AQUÍ
ciudades_calidas = [e["nombre"] for e in estaciones if e["temperatura"]>27] # Se generara la lista con compresión y se utiliza el condicional if para filtrar solo las que cumplen que son mayores a 27


print(ciudades_calidas)
````

## Ejercicio 7 — Comprensión de diccionarios (10 puntos)

Construye `poblacion_centro`: un diccionario `nombre: poblacion`, solo con las estaciones cuya `region` sea `"Centro"`, usando una comprensión de diccionarios con condición.


---
---
En el ejercicio 7 se crea un diccionario donde se guardan los nombres de cada ciudad junto con la poblacion recorriendo las estaciones en region con una compresion de diccionarios. Se utiliza if para seleccionar unicamente las ciudades que pertenecen a "Centro" en region.
---
---


````python
# ESCRIBE TU CÓDIGO AQUÍ
poblacion_centro = {e["nombre"]:e["poblacion"] for e in estaciones if e["region"]=="Centro"} #Se genera el diccionario usando {} y el formato key1:value1, y el condicional if para que solo tome las que cumplen en region "Centro"


print(poblacion_centro)
````

Construye `temperaturas_altas`: una sola lista plana con todas las temperaturas de `lecturas_semana` (sin importar de qué ciudad vinieron) que sean mayores a 30, usando **una sola comprensión anidada** (dos `for` + un `if`, en una sola expresión).

# Reto integrador (10 puntos)

Construye `resumen_norte`: una lista de cadenas con el formato `"N. nombre (temperatura°C)"`, que cumpla **todas** estas condiciones:

1. Solo incluye estaciones cuya `region` sea `"Norte"`.
2. Está ordenada de mayor a menor temperatura.
3. Está numerada empezando en 1 (el número va al inicio de cada cadena, como en el Ejercicio 1).

Debes combinar al menos tres de las herramientas vistas en esta sesión (comprensión de listas, `sorted`, `enumerate`, `zip`, `reversed`). No hay una única forma de combinarlas, pero el resultado final debe ser exactamente el mismo para todos: dos elementos, con Juárez primero.


---
---
Para el ejercicio 10 el  código primero selecciona únicamente las ciudades que pertenecen a la región Norte. Después, las ordena de la temperatura más alta a la más baja key=operator.itemgetter("temperatura") y (reverse=True). Luego enumerate les agrega un número empezando desde 1 y se crea una lista donde aparece el número{i}, el nombre de la ciudad {e['nombre']}  y su temperatura ({e['temperatura']}°C)".
---
---


````python
# ESCRIBE TU CÓDIGO AQUÍ
import operator #para poder utilizar key=operator.itemgetter("temperatura") en sorted
resumen_norte = [f"{i}. {e['nombre']} ({e['temperatura']}°C)" for i, e in enumerate(sorted([e for e in estaciones if e["region"] == "Norte"],key=operator.itemgetter("temperatura"), reverse=True), start=1)]

#Primero se ejecuta el condicional de la region  que sea "Norte"
#Posteriormente las ordena con sorted de mayor a menor gracias a reverse=True con la key de "temperatura"
#En seguida las enumera comenzando con 1 gracias a start=1
#Se crea el texto con las variables en el f-string
#Usamos los [] para la lista de compresion
print(resumen_norte)
````
