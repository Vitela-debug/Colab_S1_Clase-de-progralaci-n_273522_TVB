## Parte 1 — Vectores: perfiles de producto

Una tienda en línea representa cada producto como un vector de 3 componentes: calificación de calidad, calificación de precio, y calificación de servicio (escala 1 a 5, donde 5 es mejor). Ejecuta la siguiente celda para cargar los tres productos — **no la modifiques**.

````python
producto_A = np.array([4.5, 3.2, 5.0])
producto_B = np.array([4.0, 2.8, 4.7])
producto_C = np.array([1.0, 5.0, 1.2])

print(producto_A, producto_B, producto_C)
````

---

### Ejercicio 1 — Norma de los vectores (10 pts)

Calcula la norma de `producto_A`, `producto_B` y `producto_C` usando `np.linalg.norm()`.

````python
# Tu código aquí
norma_A = np.linalg.norm(producto_A)
norma_B = np.linalg.norm(producto_B)
norma_C = np.linalg.norm(producto_C)
print("Norma_A:",norma_A, "norma_B:",norma_B, "norma_C:",norma_C)
````

---

### Ejercicio 2 — Producto punto y ángulo (15 pts)

1. Calcula el producto punto entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`, usando `np.dot()` o `@`.
2. Calcula el ángulo (en grados) entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`.
3. Con base en los ángulos obtenidos, indica (en un comentario o celda de texto) cuál de los dos productos (`B` o `C`) es más similar a `producto_A`, y por qué el ángulo es el indicador correcto para esa comparación.

````python
# Tu código aquí
producto_punto_AB = np.dot(producto_A,producto_B) #producto = producto_A @ producto_B
producto_punto_AC = np.dot(producto_A,producto_C)
print("1. producto_punto_AB:",producto_punto_AB,"producto_punto_AC:",producto_punto_AC)
angulo_AB = np.degrees(np.arccos(producto_punto_AB/(norma_A*norma_B))) #np.degrees() para que sea en grados
angulo_AC = np.degrees((np.arccos(producto_punto_AC/(norma_A*norma_C))))
print("2. angulo_AB:",angulo_AB,"angulo_AC:",angulo_AC)

if angulo_AB < angulo_AC:
    print("3. El producto B es más similar a producto_A porque tiene un ángulo menor.")
else:
    print("3. El producto C es más similar a producto_A porque tiene un ángulo menor.")

print("El ángulo es el indicador correcto porque permite medir la similitud entre los vectores según la dirección que tienen.")
print("Mientras menor sea el ángulo entre dos vectores, más similares son sus direcciones. Además, el cálculo del ángulo normaliza")
print("las magnitudes de los vectores, por lo que la comparación no depende únicamente de qué vector tenga valores más grandes.")
print("En otras palabras: 0° = muy similares, 90° = no tienen similitud direccional, y 180° = direcciones opuestas.")
````

### Ejercicio 3 — Normalización (10 pts)

Normaliza los tres vectores (`producto_A`, `producto_B`, `producto_C`) para obtener sus vectores unitarios. Verifica, imprimiendo el resultado, que la norma de cada vector normalizado sea 1.

---
Nota: un vector unitario es un vector con norma igual a 1 — conserva la dirección del vector original, pero no su magnitud. Se obtiene dividiendo el vector entre su propia norma.

````python
# Tu código aquí
producto_A_unitario = producto_A/norma_A
producto_B_unitario = producto_B/norma_B
producto_C_unitario = producto_C/norma_C

print("Vector unitario A:",producto_A_unitario,"Vector unitario B:",producto_B_unitario,"Vector unitario C:",producto_C_unitario)

#Verificamos con la norma del vector

print("Norma producto_A_unitario:", np.linalg.norm(producto_A_unitario))
print("Norma producto_B_unitario:", np.linalg.norm(producto_B_unitario))
print("Norma producto_C_unitario:", np.linalg.norm(producto_C_unitario))
````

## Parte 2 — Matrices: mezcla de insumos

Una fábrica de jugos combina dos insumos (concentrado de fruta y agua) en distintas proporciones para dos líneas de producto. La matriz `M` representa esas proporciones (filas = línea de producto, columnas = insumo), y `P` representa un ajuste de escala para una nueva presentación. Ejecuta la siguiente celda — **no la modifiques**.

````python
M = np.array([[2, 1],
              [1, 3]])

P = np.array([[1.5, 0],
              [0, 2.0]])

print(M)
print(P)
````

### Ejercicio 4 — Multiplicación matricial (15 pts)

1. Calcula `M * P` (elemento a elemento) y `M @ P` (multiplicación matricial). Los resultados son distintos — en un comentario o celda de texto, explica brevemente por qué.
2. Verifica que las formas sean compatibles: ¿cuál sería el resultado de `P @ M`? Calcúlalo y compáralo con `M @ P` — ¿da el mismo resultado?

````python
# Tu código aquí
Elemento_Elemento = M * P
Multiplicacion_matricial = M @ P
print("1. Elemento_Elemento M*P:",Elemento_Elemento)
print("Multiplicacion_matricial M@P:",Multiplicacion_matricial)

Multiplicacion_matricial2 = P @ M
print("Verificacion de formas compartibles")
print("2.Multiplicacion_matricial P@M:",Multiplicacion_matricial2)
print("¿Da el mismo resultado?")
print("M * P==M @ P:",M * P==M @ P)
print("No, # M * P realiza multiplicación elemento a elemento, mientras que M @ P realiza multiplicación matricial, aunque M y P tienen formas compatibles para ambas operaciones. M @ P y P @ M no necesariamente son iguales porque la multiplicación matricial no es conmutativa.")
````

### Ejercicio 5 — Determinante e inversa (15 pts)

1. Calcula el determinante de `M` con `np.linalg.det()`. Confirma que `M` sí tiene inversa (determinante distinto de 0).
2. Calcula la inversa de `M` con `np.linalg.inv()`.
3. Verifica tu resultado calculando `M @ M_inv` — debe dar (aproximadamente) la matriz identidad.

````python
# Tu código aquí
determinante_M=np.linalg.det(M)
print("1. Determinante de M:", determinante_M)
inversa_M=np.linalg.inv(M)
print("2. Inversa de M:", determinante_M)
Identidad_M=M @ inversa_M
print("3. Matriz identidad de M @ inversa_M:", Identidad_M)
````

La fábrica necesita producir un lote que requiere en total 180 unidades de la línea 1 y 300 unidades de la línea 2, según las proporciones de `M`. Esto se representa como el sistema $Mx = b$, con:

### Ejercicio 7 — Síntesis: distancia y similitud (20 pts)

Retomando los tres productos de la Parte 1:

1. Calcula la **distancia euclidiana** entre `producto_A` y `producto_B`, y entre `producto_A` y `producto_C`, usando la norma de la diferencia entre los vectores (`np.linalg.norm(a - b)`).
2. Calcula la **similitud coseno** entre los mismos pares — es el cociente que ya calculaste en el Ejercicio 2 antes de aplicar `arccos()`: $\frac{a \cdot b}{\|a\| \|b\|}$.
3. Con base en ambas métricas (distancia y similitud coseno), concluye en una cadena de texto impresa cuál producto (`B` o `C`) es más parecido a `producto_A`, y si ambas métricas coinciden en la misma conclusión.

````python
# Tu código aquí
ecluidiana_AB = np.linalg.norm(producto_A - producto_B)
ecluidiana_AC = np.linalg.norm(producto_A - producto_C)
print("1. ecluidiana_AB:",ecluidiana_AB,"ecluidiana_AC:",ecluidiana_AC)

coseno_AB = np.dot(producto_A,producto_B)/(norma_A*norma_B)
coseno_AC = np.dot(producto_A,producto_C)/(norma_A*norma_C)
print("2. coseno_AB:",coseno_AB,"coseno_AC:",coseno_AC)

mas_parecido_distancia = "B" if ecluidiana_AB < ecluidiana_AC else "C"
mas_parecido_coseno = "B" if coseno_AB > coseno_AC else "C"

coinciden = mas_parecido_distancia == mas_parecido_coseno
print("3.")
print(f"Distancia A-B: {ecluidiana_AB:.4f}")
print(f"Distancia A-C: {ecluidiana_AC:.4f}")
print(f"Similitud coseno A-B: {coseno_AB:.4f}")
print(f"Similitud coseno A-C: {coseno_AC:.4f}")

print(
    f"Producto más parecido a A según distancia: {mas_parecido_distancia}. "
    f"Según similitud coseno: {mas_parecido_coseno}. "
    f"Ambas métricas coinciden: {'Sí' if coinciden else 'No'}.")
````
