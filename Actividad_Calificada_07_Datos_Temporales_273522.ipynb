2. Cuenta y muestra los viajes cuyo `pickup` **no** está en marzo de 2019.

Finalmente, `len()` cuenta cuántos viajes cumplen esas condiciones. ### 1.3 Cambio de horario (5 pts)

El 10 de marzo de 2019, en Nueva York, las horas entre 02:00 y 02:59 **no existieron**.

1. Filtra los viajes con `pickup` el 10 de marzo y cuenta cuántos hay en cada hora (`.dt.hour` y `groupby()` o `value_counts()`). ¿Hay viajes a las 2:00?
2. Localiza `pickup` en la zona `'America/New_York'` con `tz_localize()` usando `nonexistent='NaT'` y cuenta cuántos valores quedan como `NaT`.

```python
# Escribe tu código aquí

viajes_10_marzo = taxis[taxis['pickup'].dt.date == pd.Timestamp('2019-03-10').date()] #Filtra del pickup y los que son del 10 de marzo
viajes_por_hr = viajes_10_marzo['pickup'].dt.hour.value_counts().sort_index() #.dt.hour extrae la hora del viaje, .value_counts() cuenta los viajes por hora y .sort_index() las ordena
print(viajes_por_hr)

if 2 in viajes_por_hr:
  print('Hay viajes a las 2:00-2:59') #Comprobamos con un if que a las 2 no se tienen registrados viajes
else:
  print('No hay viajes a las 2:00-2:59')

registros_ZONE = taxis['pickup'].dt.tz_localize('America/New_York',nonexistent='NaT') #.dt.tz_localize()  asigna una zona horaria a las fechas y nonexistent='NaT' cuando no existe la hora lo asigna
print('Valores con NaT en pickup', registros_ZONE.isna().sum()) #Cuenta los NaT


```

Luego se utiliza un `if` para comprobar si existe la hora 2, indicando si hubo o no viajes entre las 2:00 y 2:59. Finalmente, .`dt.tz_localize()` asigna la zona horaria America/New_York a pickup y, con `nonexistent='NaT'`, convierte en NaT las horas que no existen, `.isna().sum() `cuenta cuántos registros quedaron como NaT.

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# Escribe tu código aquí
import numpy as np
#1
velocidad_mph = taxis['distance'] / (taxis['duracion_min'] / 60) #Calculamos la velocidad
taxis['velocidad_mph'] = velocidad_mph.where(taxis['duracion_min'] > 0, np.nan) #se restringe con where y se condicionan a NaN los que son menores a 0
print('Conversiones a NaN:',taxis['velocidad_mph'].isna().sum())

#2
viajes_velocidad_60 = taxis[taxis['velocidad_mph'] > 60] #Viajes con velocidad mayor a 60 con condicional >
print('Viajes con velocidad mayor a 60 mph:', len(viajes_velocidad_60)) #contar los viajes con velocidades mayores a 60

#3
taxis_odenado = taxis.sort_values(by='velocidad_mph', ascending=False) #Ordenar los viajes por velocidad de mayor a menor

print('\n5 viajes con mayor velocidad:')
print(taxis_odenado[['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']].head(5)) #mostar ciertas columnas de los datos ordenados por velocidades

```

---
---
* El código del ejercicio 2.3 calcula la velocidad de cada viaje en mph, usando la distancia (distance) y la duración (duracion_min). Con .where() controla los casos donde la duración no es válida. Después, con filtros ([] y >) saco los viajes que superan las 60 mph y con len() cuento cuántos hay. Finalmente, uso .sort_values() para ordenar los viajes de mayor a menor velocidad y .head(5) para mostrar los 5 más rápidos, seleccionando solo algunas columnas.

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Escribe tu código aquí

taxis['hora'] = taxis['pickup'].dt.hour #Extraemos la hora de pickup
taxis['dia_num'] = taxis['pickup'].dt.dayofweek #Extraemos el día de la semana de pickup
taxis['dia_nombre'] = taxis['pickup'].dt.day_name() #Extraemos el nombre del día de pickup
taxis['fin_semana'] = taxis['dia_num'].isin([5, 6]) #Extraemos si es fin de semana, sábado (5) y domingo (6)
taxis['fecha'] = taxis['pickup'].dt.date #Extraemos la fecha de pickup

taxis['fecha_dro'] = taxis['dropoff'].dt.date
taxis['cruza_medianoche'] = taxis['fecha'] !=  taxis['dropoff'].dt.date #Extraemos si cruza la medianoche


print(taxis[['hora', 'dia_num', 'dia_nombre', 'fin_semana', 'fecha', 'cruza_medianoche', 'fecha_dro']].head(10)) #mostrar las primeras 10 filas


print('\n Viajes que cruzan la medianoche:', taxis['cruza_medianoche'].sum()) #Contar viajes con criza a medianoche
```

---
---
* En el código del ejercicio 3.1 se utilizó `taxis['pickup'].dt `para extraer información de la fecha y hora del pickup. Con `.dt.hour` obtengo la hora, con .`dt.dayofweek` saco el número del día de la semana y con .dt.day_name() obtengo su nombre. Después uso `.isin([5, 6]) `para identificar si el viaje fue en fin de semana, considerando 5 como sábado y 6 como domingo. Con .`dt.date `extraigo solamente la fecha del pickup y del dropoff. Finalmente, con `!=` comparo ambas fechas para saber si el viaje cruzó la medianoche, y con `.head(10)` muestro las primeras 10 filas. También utilizo `.sum()` sobre cruza_medianoche para contar cuántos viajes terminaron en un día diferente al que comenzaron.

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Escribe tu código aquí

# Condiciones para excluir viajes
cond_duracion = taxis['duracion_min'] <= 0 #menores o iguales que cero
cond_velocidad = taxis['velocidad_mph'] > 60 #velocidades que superar las 60 mph
cond_fecha = ~taxis['pickup'].between('2019-03-01', '2019-03-31 23:59:59') #Excluir los datos fuera de marzo 2019

# Contar viajes excluidos por cada condición
print("Viajes excluidos por duración:", cond_duracion.sum()) #contar los datos excluido por duracion
print("Viajes excluidos por velocidad:", cond_velocidad.sum()) #contar los datos excluido por velocidad
print("Viajes excluidos por fecha:", cond_fecha.sum()) #contar los datos excluido por fecha

# Crear el dataset limpio
taxis_limpio = taxis[ ~(cond_duracion | cond_velocidad | cond_fecha)].copy() #para quitar los datos que no se quieren

# Contar el total de viajes excluidos
print("Total de viajes excluidos:", len(taxis) - len(taxis_limpio)) #Total de excluidos

# Verificar que taxis_limpio cumple las condiciones
print("\nVerificación:")
print("Duraciones <= 0:", (taxis_limpio['duracion_min'] <= 0).sum()) #Contar duraciones menores de 0 en taxis_limpio
print("Velocidades > 60 mph:", (taxis_limpio['velocidad_mph'] > 60).sum()) #contar velocidades mayores de 60 en taxis_limpio
print("Pickups fuera de marzo 2019:", (~taxis_limpio['pickup'].between('2019-03-01', '2019-03-31 23:59:59')).sum()) #contar fechas fuera de marzo en datos limpio
```

Con `.sum()` cuento cuántos viajes se excluyen por cada condición. Después se usa `taxis_limpio = taxis[~(cond_duracion | cond_velocidad | cond_fecha)].copy() `para crear el dataset limpio, quedándo solo con los viajes que cumplen todas las condiciones. Luego se calculo el total de viajes excluidos con `len(taxis) - len(taxis_limpio)` y finalmente se volvieron a contar las condiciones en taxis_limpio para verificar que ya no haya duraciones menores o iguales a 0, velocidades mayores a 60 mph ni pickups fuera de marzo de 2019.
