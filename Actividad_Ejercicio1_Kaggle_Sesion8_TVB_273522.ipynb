**Paso 1 y 2 — Configurar credenciales y descargar el dataset.**

```python
# Instala la librería kaggle (si no está instalada ya)
!pip install kaggle --quiet

import os

# Carga el token desde los Secretos de Colab (Paso 1)
try:
    from google.colab import userdata #cceder a datos almacenados como secretos
    os.environ["KAGGLE_API_TOKEN"] = userdata.get("KAGGLE_API_TOKEN") #obtiene el token de Kaggle que está guardado en los secretos de Google Colab
    print("Token de Kaggle cargado")
except ImportError: #manejo de errores
    print("Esta celda está diseñada para ejecutarse en Google Colab con el Secreto configurado")

# Completa el identificador del dataset y descárgalo en datos_drug/ (Paso 2)
dataset_id = "prathamtripathi/drug-classification"  # <- completa aquí #Se agergo el identificador correcto del dataset.

!kaggle datasets download -d {dataset_id} -p datos_drug/ --unzip
```

---
---
Para el paso 1 y 2 el código sirve para descargar un dataset de Kaggle desde Google Colab de forma segura.  
* !pip install kaggle --quiet instala la herramienta de Kaggle para poder utilizarla.
* Con el try se intenta acceder a los Secretos de Google Colab mediante userdata para obtener el token de Kaggle y guardarlo como una variable llamada KAGGLE_API_TOKEN, lo que permite comprobar que se tiene el permiso para descargar el dataset sin escribir el token directamente en el código.
  *  El print muestra un mensaje para confirmar que el token fue cargado ("Token de Kaggle cargado") y el except muestra un aviso si no se está utilizando correctamente Google Colab.
*  Luego, dataset_id guarda el identificador del dataset que se quiere descargar (prathamtripathi/drug-classification).
*  Por su parte !kaggle datasets download -d {dataset_id} -p datos_drug/ --unzip descarga el dataset desde Kaggle y lo guarda en la carpeta datos_drug
*  Lo descomprime automáticamente utilizando --unzip.

**Paso 3 — Cargar el archivo descargado en un DataFrame.**

```python
import glob #glob → encuentra los archivos
import pandas as pd #pandas abre y manipula los datos

archivos_csv = glob.glob("datos_drug/*.csv")
print(archivos_csv)

# Completa: carga el primer archivo de archivos_csv en un DataFrame llamado df_drug
df_drug = pd.read_csv(archivos_csv[0]) #carga el primer archivo de archivos_csv, lee y lo converte en un DataFrame llamado df_drug
```

---
---
Para el paso 3 el código importa glob para buscar archivos y pandas para trabajar con los datos. Se usa  glob.glob() para buscar los archivos CSV dentro de la carpeta datos_drug y se guardan los archivos encontrados en archivos_csv. Luego se muestra cuales archivos encontró.
*  El  pd.read_csv() se utiliza junto con archivos_csv[0] para tomar el primer archivo de la lista, leerlo y convertirlo en un DataFrame llamado df_drug.

**Paso 4 — Verificar el resultado.**

```python
# Completa: imprime shape, columns, dtypes y valores faltantes de df_drug,
# siguiendo el mismo patrón usado en los Bloques 1 y 2.
print("Dimensiones:", df_drug.shape)
print("------------------------------------------------------------")
print("------------------------------------------------------------")
print("Columnas:", df_drug.columns.tolist())
print("------------------------------------------------------------")
print("------------------------------------------------------------")
print("Dtypes:")
print(df_drug.dtypes)
print("------------------------------------------------------------")
print("------------------------------------------------------------")
print("Dtypes con df_drug.info:")
df_drug.info()
print("------------------------------------------------------------")
print("------------------------------------------------------------")
print("Valores faltantes por columna:")
print(df_drug.isna().sum())
```

---
---
El código de la parte 4 sirve para revisar la información básica del DataFrame df_drug:
*  df_drug.shape para conocer cuántas filas y columnas tiene
*  df_drug.columns.tolist() para obtener los nombres de las columnas
*  df_drug.dtypes para identificar qué tipo de datos tiene cada columna y df_drug.info() para obtener un resumen general de la estructura del DataFrame (**se utilizan ambos para observar las diferencias y como se organizan los datos**)
*  df_drug.isna().sum() para saber cuántos valores faltantes existen en cada columna

**Paso 5 — Ficha de procedencia.** Completa cada valor del diccionario (usa el mismo patrón visto en
los Bloques 1 y 2).

```python
from datetime import datetime, timezone

ficha_drug = {
    "dataset": "prathamtripathi/drug-classification", #nombre del dataset
    "fuente": "Proveniente de Kaggle",  #Origen de dataset
    "fecha_descarga_utc": datetime.now(timezone.utc).isoformat(), #para guardar la fecha y hora exactas en que descargaste el dataset,
    "licencia": "CC0: Public Domain",  # revisa la página del dataset en kaggle.com para completar este dato ( link: https://www.kaggle.com/datasets/prathamtripathi/drug-classification)
}

ficha_drug


```

---
---
El código de la parte 5 sirve para crear una ficha con la información principal del dataset que se descargó desde Kaggle.
*  Se importa datetime y timezone para poder obtener la fecha y hora actual (cada vez que se corra el programa se actualizara). Después creo un diccionario llamado ficha_drug, donde guardo diferentes datos del dataset.
* En "dataset" se pone el nombre del dataset -> prathamtripathi/drug-classification.
* En "fuente" se indica que el dataset proviene de Kaggle.
* En "fecha_descarga_utc" se guarda automáticamente la fecha y hora en la que se ejecuta el código. UTC para tener una hora estándar y .isoformat() para guardar la fecha en un formato ordenado.
* En "licencia" se indica que el dataset tiene una licencia CC0: Public Domain, que es la licencia que aparece en la página de Kaggle (https://www.kaggle.com/datasets/prathamtripathi/drug-classification)
* Y con ficha_drug, se observa toda la información que se guardo en el diccionario
