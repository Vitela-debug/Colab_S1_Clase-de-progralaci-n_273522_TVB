## Actividad 2 — Ajuste de tipos: fechas con formato mixto

*Dataset: Netflix*

La columna `date_added` de `df_netflix` mezcla formatos reales: la mayoría son `"14-Aug-20"`, pero un grupo minoritario llega como `" August 4, 2017"` (con espacio inicial). Conviértela a tipo fecha usando `pd.to_datetime(..., format='mixed')`, que resuelve ambos formatos en la misma columna. Verifica con `.dtypes` y confirma cuántos valores nulos quedan después de la conversión (compara contra los nulos que ya traía antes de convertir).

```python
# Tu código aquí
# Revisar los tipos de datos antes de hacer la conversión lo esperado son objects
print("Tipos de datos antes de la conversión:", df_netflix['date_added'].dtypes)

# Revisar cuántos valores nulos tiene date_added antes de convertirla
nulos_antes = df_netflix['date_added'].isna().sum()
print("Nulos antes:", nulos_antes)

#Visualizar los nulos antes
df_netflix.loc[df_netflix['date_added'].isna(), 'date_added']

```

```python
# Convertir date_added a formato fecha, permitiendo formatos mixtos (fechas escritas de diferentes maneras)
df_netflix['date_added'] = pd.to_datetime(df_netflix['date_added'], format='mixed')

# Revisar los tipos de datos después de la conversión lo esperado es datetime
print("Tipos de datos despues de la conversión:", df_netflix['date_added'].dtypes)

# Revisar cuántos valores nulos quedaron después de la conversión
nulos_despues = df_netflix['date_added'].isna().sum()
print("Nulos después:", nulos_despues)

#Visualizar los nulos despues de la conversion
df_netflix.loc[df_netflix['date_added'].isna(), 'date_added']

```

```python
print("Los valores nulos antes y despues son iguales:", nulos_antes == nulos_despues )
```

---
---
* El código de la actividad 2 primero revisa el tipo de dato de date_added y la cantidad de valores nulos que tenía antes de la conversión mediante `df_netflix['date_added'].dtypes` y `df_netflix['date_added'].isna().sum()` para contar los nulos. Después se utiliza `pd.to_datetime()` con `format='mixed'` porque la columna contiene fechas en diferentes formatos, lo que permite que pandas reconozca diferente tipo de fechas, de tal forma que, la columna cambió de object a datetime64[ns]. Los valores nulos se mantuvieron en 10 porque ya estaban vacíos antes de la conversión y `pd.to_datetime()` no rellena estos valores, solamente convierte las fechas que sí tienen información. Por lo tanto, que los nulos sean los mismos antes y después indica que no se generaron nuevos valores nulos durante la conversión.

## Actividad 5 — Completitud como porcentaje

*Dataset: Netflix*

Con los mismos nulos de la Actividad 4, calcula la completitud en porcentaje por columna: `(1 - nulos / total_filas) * 100`. ¿Qué columna tiene la completitud más baja? Escribe la respuesta en una línea.

```python
# Tu código aquí
# len(df_netflix) da el número total de filas -- lo usamos como base para calcular el porcentaje.
total_registros = len(df_netflix)
# (1 - proporción_de_nulos) * 100 = porcentaje de valores presentes.
completitud = (1 - df_netflix.isnull().sum() / total_registros) * 100
completitud

```

```python
columna_minima = completitud.idxmin()  #idxmin() busca el nombre de la columna que tiene el valor más pequeño.
print("La columna con menor completitud es:", columna_minima) #para sacar la columna que tiene más valores faltantes proporcionalmente
print("Su completitud es:", completitud.min(), "%")  #para sacar el valor con  la columna que tiene más valores faltantes proporcionalmente en terminos de %
```

---
---
* El código de la Actividad 5 primero calcula el número total de registros de la base de datos con `len(df_netflix)` para tener una cantidad de referencia y poder calcular los porcentajes.
8 `df_netflix.isnull().sum()` cuenta los valores faltantes de cada columna y con la fórmula `(1 - nulos / total_filas) * 100` se calcula el porcentaje de datos que sí están completos en cada una.
* Con completitud se puede  ver qué tan completa está cada columna.
* Luego, con `completitud.idxmin()` se identifica el nombre de la columna que tiene el porcentaje de completitud más bajo, es decir, la que tiene proporcionalmente más datos faltantes. Finalmente, con `completitud.min() ` se obtiene el porcentaje mínimo de completitud para saber qué tan completa está esa columna.
