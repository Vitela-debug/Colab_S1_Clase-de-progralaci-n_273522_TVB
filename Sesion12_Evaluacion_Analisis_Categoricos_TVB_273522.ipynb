## Actividad 3 — Alta cardinalidad (25 pts)

```python
# 3.1 — Calcula .nunique() para TODAS las columnas del dataset (no solo las categóricas)
# y la razón (valores únicos / total de filas) para cada una.

columnas = df.columns.tolist()

for col in columnas:
    n_unicos = df[col].nunique()
    ratio = n_unicos / len(df)
    print(f'{col:12s} -> {n_unicos:5d} valores únicos  ({ratio:.1%} del total de filas)')

#Este código sirve para revisar cuántos valores diferentes tiene cada columna y
#que porcentaje representan del total de filas. df.columns.tolist() obtiene la
#lista de columnas, nunique() cuenta los valores únicos y ratio calcula qué porcentaje representan.
```

**3.2 — Conclusión (responde aquí en Markdown):**

¿Qué columna(s) tienen alta cardinalidad? Para la columna con mayor cardinalidad: ¿por qué nunca deberías usarla como variable predictora en un modelo, incluso si la codificaras? (relaciona tu respuesta con lo discutido en clase sobre identificadores únicos)

_Tu respuesta:_

La columna con mayor cardinalidad es  customerID, porque tiene 7043 valores únicos, es decir, un identificador diferente para cada cliente.

No debería usarse como variable predictora porque es un identificador único y no contiene información útil sobre las características del cliente. Aunque se codificara, el modelo podría aprender los identificadores en lugar de patrones reales de los datos, lo que podría causar sobreajuste y no ayudaría a predecir correctamente nuevos clientes. La segunda columna con mayor cardinalidad es TotalCharges, con 6531 valores únicos. A diferencia de customerID, no es un identificador, sino una variable que representa el total cobrado al cliente. Por eso, sí puede contener información útil para el modelo y no necesariamente debe eliminarse por tener alta cardinalidad.

**3.3 — Agrupación "Top 10 + Otros"**

En clase agrupaste `country` de Netflix Titles en sus 10 categorías más frecuentes + `'Otros'`, reduciendo su cardinalidad. Aplica la misma técnica aquí sobre la columna de mayor cardinalidad que identificaste en 3.1 (pista: `.value_counts().head(10)`, luego `.where()` + `.isin()`, igual que en el notebook de clase).

```python
# Aplica el agrupamiento Top 10 + Otros sobre la columna de mayor cardinalidad

top10 = df['TotalCharges'].value_counts().head(10)
categorias_top10 = top10.index.tolist()
df['TotalCharges_agrupado'] = df['TotalCharges'].where(
    df['TotalCharges'].isin(categorias_top10), 'Otros'
)
df['TotalCharges_agrupado'].value_counts()

#El código del ejercicio 3.3 sirve para agrupar los valores de TotalCharges:
#value_counts().head(10) encuentra los 10 valores más repetidos,
# where() mantiene esos 10 y cambia los demás por "Otros", y value_counts()
#muestra cuántos quedaron en cada grupo.

```

**3.4 — Conclusión (responde aquí en Markdown):**

Después de agrupar, ¿la columna resultante te parece útil para un modelo? Compara este caso con el de `country` en el notebook de clase: ¿por qué agrupar en "Top 10 + Otros" funciona bien para una variable como `country`, pero no resuelve el problema real de la columna que agrupaste aquí?

_Tu respuesta:_

La columna TotalCharges_agrupado no me parece muy útil para el modelo, porque casi todos los valores quedaron dentro de la categoría "Otros" y se perdió mucha información.

A diferencia de country, donde agrupar como "Top 10 + Otros" permite reducir muchas categorías diferentes manteniendo las más importantes, TotalCharges representa una cantidad numérica. Agruparla de esta manera hace que se pierda información sobre cuánto ha pagado realmente cada cliente. Por eso, en este caso sería mejor mantener TotalCharges como variable numérica en lugar de agruparla.

## Actividad 4 — Tipos de dato (30 pts)

```python
df['TotalCharges'].dtype
```

**4.1 — Investiga (responde en Markdown):**

`TotalCharges` contiene valores numéricos (montos en dólares), pero pandas la cargó como `object`, no como `float`. Investiga por qué — revisa si hay algún valor que no se vea como un número normal.

_Tu respuesta:_

En este dataset, el problema es que TotalCharges tiene algunos valores vacíos (se puede ver en la tabla de top 10 que se tienen valores vacios). Esos valores no se pueden interpretar como números, por lo que pandas carga toda la columna como object.

```python
# 4.2 — Corrige el tipo de TotalCharges.
# Pista: pd.to_numeric() con el parámetro errors= te puede ayudar a identificar
# o manejar los valores problemáticos que encontraste en 4.1.

df["TotalCharges"]=pd.to_numeric(df["TotalCharges"], errors="coerce") #Se corrigen los valores vacios
df["TotalCharges"].dtype #se comprueba que esta bien el tipo de dato


#El código sirve para convertir TotalCharges a números, reemplazando los valores problemáticos por NaN.
#Después, dtype comprueba que la columna tenga el tipo de dato correcto
```

```python
# 4.3 — Convierte a category las columnas categóricas que, según lo que calculaste
# en la Actividad 3, tengan cardinalidad baja y valores fijos.
# Verifica con .dtypes que el cambio se aplicó correctamente.

#sabemos que columnas_categoricas son las que se verifico que eran tipo "object"
#asi que se aplicara el cambio ahi y se omitiran los que sabemos que tienen alta
#cadinalidad como customer ID

for columna in columnas_categoricas:
  if columna not in ['customerID', 'TotalCharges','TotalCharges_agrupado']: #Para no cambiar los identificadores de alta cardinalidad
        df[columna] = df[columna].astype('category')



print(df.dtypes)

#El código sirve para cambiar las columnas categóricas de tipo object a category,
#lo que ayuda a manejar mejor los datos categóricos. Y if evita cambiar columnas como customerID
# y TotalCharges, porque no son adecuadas para este cambio.
#El df.dtypes comprueba que las columnas se hayan cambiado correctamente
```
